# Products table

```python
df1 = pd.read_excel("Indian_cleaned_data.xlsx",sheet_name="products")
df1
```

## Checking outliers

```python
df1.select_dtypes(include='number').plot(
    kind='box',
    subplots=True,
    layout = (-1,3),
    figsize=(15,8)
)
plt.show()
```

```python
df1.Selling_Price_INR.max(),df1.Selling_Price_INR.min()
```

```python
df1.Cost_INR.max(),df1.Cost_INR.min()
```

```python
df1.columns
```

```python
# check actual outliers
def detect_out():
    data = df1.select_dtypes(include='number')

    for i in data:
        data=df1[i].dropna()
        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = data[(data < lower) | (data > upper)]

        print("\ncolumn", i)
        print("no_outlier :", len(outliers))
        
        if len(outliers)>0:
            print(outliers)
detect_out()
```

### Conclusion : Cost_INR and Selling_Price_INR can be valid.

# Order table : No numeric columns

```python
df5 = pd.read_excel("Indian_cleaned_data.xlsx",sheet_name="orders")
df5
```

```python
df2 = pd.read_excel("Indian_cleaned_data.xlsx",sheet_name="order_items")
df2
```

### Checking outlier

```python
df2.select_dtypes(include='number').plot(
    kind='box',
    subplots=True,
    layout = (-1,3),
    figsize=(15,8)
)
plt.show()
```

```python
df2['Discount_Pct'].max(),df2['Discount_Pct'].min()
```

```python
df2[df2['Discount_Pct'] < 0]
```

```python
df2[df2['Discount_Pct'] > 50]
```

#### All the column outlier are genuine but discount_pct has negative value which can't be consider as valid so i will replace with median 

```python
df2.columns
```

```python
# Median of valid discount values
median_discount = df2.loc[df2['Discount_Pct'] >= 0,'Discount_Pct'].median()

# Replace invalid negative discounts
df2.loc[df2['Discount_Pct'] < 0, 'Discount_Pct'] = median_discount
```

# Merge into one excel file
