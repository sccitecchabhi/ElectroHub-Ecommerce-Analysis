# Time series

```python
import pandas as pd

df = pd.DataFrame({
    "date": ["2026-06-01", "2026-06-02","2026-06-03","2026-06-04",
        "2026-06-05", "2026-06-06", "2026-06-07"],
    "sales": [100, 120, 150, 130, 170, 160, 180]  })
```

```python
df["sales"].rolling(window=3).mean() # gives average on row 3 of previous 2 + row itself on each row
df["previous_day_sales"] = df["sales"].shift(1) # gives the previous value just like lag() in sql
df.asfreq("D")  #  fill the Nan in missing dates 

df.resample("D").sum() # groupby and gives the sum of daily
df.resample("D").sum()    # Daily
df.resample("W").sum()    # Weekly
df.resample("M").sum()    # Monthly
df.resample("Y").sum()    # Yearly
```

```python
1. Fixed Value Add Karna
df['country'] = 'India'

city	country
Delhi	India

2. Serial Number Column
df['S_No'] = range(1, len(df)+1)

3. Condition Lagana (np.where)
Sales > 20000

import numpy as np

df['high_sales'] = np.where(df['sales'] > 20000,
                            'Yes',
                            'No')

4. Multiple Conditions
df['performance'] = np.where(df['sales'] > 50000,
                             'Excellent',
                             'Average')

5. Nested Conditions
df['category'] = np.where(df['sales'] > 50000,
                          'High',
                          np.where(df['sales'] > 20000,
                                   'Medium',
                                   'Low'))

6. Mathematical Calculation
Profit Margin
df['profit_margin'] = (df['profit']/df['sales'])*100

Industry me bahut common.

7. Age Group Column
df['age_group'] = np.where(df['age'] < 30,
                           'Young',
                           'Senior')

8. String Condition
df['is_delhi'] = np.where(df['city'] == 'Delhi',
                          'Yes',
                          'No')

9. Multiple Cities
df['metro_city'] = np.where(
    df['city'].isin(['Delhi','Mumbai','Bangalore']),
    'Metro',
    'Non Metro'
)


10. Missing Value Flag
df['missing_city'] = np.where(
    df['city'].isna(),
    'Missing',
    'Present'
)

11. Outlier Flag
df['outlier'] = np.where(
    (df['sales'] > upper_fence) |
    (df['sales'] < lower_fence),
    'Yes',
    'No'
)

12. Date Based Column
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month_name()

13. Discount Column
df['discount'] = df['sales'] * 0.10

14. Bonus Column
df['bonus'] = np.where(df['profit'] > 10000,
                       5000,
                       1000)

15. Rank Column
df['sales_rank'] = df['sales'].rank(ascending=False)


16. Group-wise Calculation
City Wise Average Sales
df['city_avg_sales'] = df.groupby('city')['sales'].transform('mean')

Very important industry question.

17. Top Customer Flag
top_sales = df['sales'].quantile(0.90)

df['top_customer'] = np.where(
    df['sales'] >= top_sales,
    'Yes',
    'No'
)


18. Custom Function
def classify(x):
    if x > 50000:
        return 'High'
    elif x > 20000:
        return 'Medium'
    else:
        return 'Low'

df['sales_category'] = df['sales'].apply(classify)


Interview/Industry Me Sabse Important
np.where()
.isin()
.apply()
.map()
groupby().transform()
Date columns (dt.year, dt.month)
Mathematical columns
Outlier flags
Missing value flags
Ranking columns

Agar ye 10 patterns aate hain, to DA/DS me 80-90% column creation wali real-world requirements handle kar loge.
```
